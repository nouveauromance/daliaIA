### **Importando Bibliotecas e Materiais:**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys
import boxplot as bp
import pingouin as pg
import itertools
from scipy.stats import mannwhitneyu
```

### **Abrindo Dataset:**

```python
df = pd.read_csv("../data/raw/Maternal Health Risk Data Set.csv")

df.head(10)
```

### **Iniciando Limpeza:**

```python
df_clean = df.drop_duplicates().copy()

print("Antes:", df.shape)
print("Depois:", df_clean.shape)
```

### **Investigando Anomalia Pós - Limpeza:**

```python
df_clean[df_clean["HeartRate"] < 40]
```

```python
df_clean["HeartRate"].describe()
```

**Eliminando Registro de Anomalia:**

```python
df_clean = df_clean[df_clean["HeartRate"] >= 40].copy()

print("Após tratamento do HeartRate:", df_clean.shape)
```

```python
df_clean["HeartRate"].describe()
```

### **Salvando Dataset Pós - Limpeza:**

```python
os.makedirs("../data/processed", exist_ok=True)

df_clean.to_csv("../data/processed/maternal_health_clean.csv", index=False)

print("Dataset limpo salvo com sucesso!")
print("Dimensões:", df_clean.shape)
```
